## Feature Engineering

# Missing Values

### Categorical Variables

```python
features_nan = [feature for feature in data.columns if data[feature].isnull().sum()>1 and data[feature].dtypes=='O']
for feature in features_nan:
    print("{}: {}% missing values".format(feature, np.round(data[feature].isnull().mean(),2)*100))
```

```python
def replace_cat_feature(data, features_nan):
    data1 = data.copy()
    data1[features_nan]=data1[features_nan].fillna('Missing')
    return data1
data = replace_cat_feature(data, features_nan)
data[features_nan].isnull().sum()
```

```python
data.head()
```

```python
categorical_features = [feature for feature in data.columns if data[feature].dtypes == 'O']
categorical_features
```

```python
for feature in categorical_features:
    temp = data.groupby(feature)['Id'].count()/len(data)
    temp_df = temp[temp>0.01].index
    data[feature] = np.where(data[feature].isin(temp_df),data[feature], 'Rare_var')
```

```python
data.head()
```

### Numerical Variables

```python
numerical_with_nan = [feature for feature in data.columns if data[feature].isnull().sum() and data[feature].dtypes!='O']
for feature in numerical_with_nan:
    print('{}: {}% missing value'.format(feature, np.round(data[feature].isnull().mean(),3)*100))
```

```python
# replacing the missing values in the numerical dataset
for feature in numerical_with_nan:
    # we will replace by using median since there are outliers
    median_value = data[feature].median()
    # create a new feature to capture nan values
    data[feature+'nan']=np.where(data[feature].isnull().isna(),1,0)
    data[feature].fillna(median_value, inplace=True)
data[numerical_with_nan].isnull().sum()
```

```python
data.head()
```

```python
num_features = ['LotFrontage','LotArea','1stFlrSF','GrLivArea']
for feature in num_features:
    data[feature]=np.log(data[feature])
```

### Temporal Variables (Date Time)

```python
# converting the date time values in year to the number of years, calculated by taking the difference with 'YrSold'
for feature in ['YearBuilt','YearRemodAdd','GarageYrBlt']:
    data[feature]=data['YrSold']-data[feature]
```

```python
data.head()
```

```python
# encoding the categorical varaibles
```

```python
for feature in categorical_features:
    labels_ordered = data.groupby([feature])['GarageArea'].mean().sort_values().index
    labels_ordered = {k:i for i,k in enumerate(labels_ordered,0)}
    data[feature] = data[feature].map(labels_ordered)
```

```python
data.head()
```

### Feature Scaling

```python
feature_scale = [feature for feature in data.columns if feature not in ['Id']]
scaler = MinMaxScaler()
scaler.fit(data[feature_scale])
```

```python
scaler.transform(data[feature_scale])
```

```python
data = pd.concat([data['Id'].reset_index(drop=True),
                  pd.DataFrame(scaler.transform(data[feature_scale]), columns=feature_scale)],
                 axis=1)
```
